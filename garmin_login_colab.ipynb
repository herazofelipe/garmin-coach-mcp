# 🔑 Generar token de Garmin (para el MCP)

Ejecuta las celdas **en orden** (botón ▶️ a la izquierda de cada una).
Funciona desde el celular. Maneja MFA. Al final obtienes el valor `GARMIN_TOKENS`.

El token dura ~1 año y se refresca solo; solo repites esto si caduca.

```
!pip install -q garminconnect
```

```
from garminconnect import Garmin
from getpass import getpass

email = input('Email de Garmin Connect: ').strip()
password = getpass('Contrasena (no se mostrara): ')

g = Garmin(email=email, password=password,
           prompt_mfa=lambda: input('Codigo MFA (si lo pide): ').strip())
g.login()
print('Sesion OK como:', g.get_full_name())

tokens = g.client.dumps()
with open('garmin_tokens.txt','w') as f:
    f.write(tokens)

print('\n===== COPIA TODO LO SIGUIENTE como valor de GARMIN_TOKENS =====\n')
print(tokens)
print('\n===== (tambien se guardo en garmin_tokens.txt) =====')
```
